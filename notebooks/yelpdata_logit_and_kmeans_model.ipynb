Research Q: Can business attributes help classify whether a restuarant has a high recorded average Yelp Rating? 

- High rating = 1:recorded average rating is at least 4.0 stars.
- Not high rating = 0:recorded average rating is below 4.0 stars.

This cutoff is an operational choice for this model, not a Yelp-defined category. The original half-star `stars` value is retained for reference, but the classifier predicts the binary outcome.

The notebook uses the prepared **Restaurants development** data created by the EDA notebook. It creates a validation split from development data and leaves the separate project holdout untouched for a final, coordinated comparison across group models.

```python
#packages and project data 

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay,
    classification_report, RocCurveDisplay
)

# Works when the notebook is run from mlproj or a subfolder such as notebooks/.
ROOT = next(
    (p for p in [Path.cwd(), *Path.cwd().parents]
     if (p / "data/02_processed/restaurants_development.csv").exists()),
    None
)
if ROOT is None:
    raise FileNotFoundError(
        "Could not find data/02_processed/restaurants_development.csv. "
        "Run the EDA notebook first and check that this notebook is inside mlproj."
    )

DATA_PATH = ROOT / "data/02_processed/restaurants_development.csv"
yelpdf = pd.read_csv(DATA_PATH)
print(f"Loaded {len(yelpdf):,} development restaurants from {DATA_PATH}")
print(yelpdf.shape)
```

```python
#inspecting prepared data and defining outcome
print("Columns:")
print(yelpdf.columns.tolist())

print("\nOriginal star-rating distribution:")
print(yelpdf["stars"].value_counts().sort_index())

# Binary outcome: 1 = at least 4.0 stars; 0 = below 4.0 stars.
RATING_CUTOFF = 4.0
yelpdf = yelpdf.loc[yelpdf["stars"].between(1, 5)].copy()
yelpdf["high_rating"] = (yelpdf["stars"] >= RATING_CUTOFF).astype(int)

print("\nBinary outcome counts:")
print(yelpdf["high_rating"].value_counts().rename(index={0: "Below 4.0", 1: "4.0 or higher"}))
print("\nBinary outcome proportions:")
print(yelpdf["high_rating"].value_counts(normalize=True).sort_index().round(3))
```

```python
# selecting predictors 
numeric_features = [
    "latitude", "longitude", "category_count", "listed_days", "weekly_listed_hours"
]

categorical_features = [
    "state", "price_range", "takeout", "delivery", "outdoor_seating",
    "reservations", "alcohol", "wifi", "noise_level"
]

feature_columns = numeric_features + categorical_features
missing_columns = [c for c in feature_columns + ["high_rating"] if c not in yelpdf.columns]
if missing_columns:
    raise KeyError(f"Expected columns are missing from the prepared dataset: {missing_columns}")

X = yelpdf[feature_columns].copy()
y = yelpdf["high_rating"].copy()

print(f"Predictors: {len(feature_columns)} ({len(numeric_features)} numeric, {len(categorical_features)} categorical)")
print("\nMissingness in selected predictors (%):")
print((X.isna().mean() * 100).round(1).sort_values(ascending=False))
```

```python
# 80/20 validation split 
X_train, X_valid, y_train, y_valid = train_test_split(
    X, y,
    test_size=0.20,
    random_state=6021,
    stratify=y
)

print(f"Training rows:   {len(X_train):,}")
print(f"Validation rows: {len(X_valid):,}")
print(f"High-rating share, training:   {y_train.mean():.3f}")
print(f"High-rating share, validation: {y_valid.mean():.3f}")
```

```python
# logistic regression pipeline (preventing data leakage)
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=50
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

logit_model = Pipeline([
    ("preprocessor", preprocessor),
    ("logistic_regression", LogisticRegression(
        max_iter=2000,
        C=1.0,
        solver="lbfgs"
    ))
])

logit_model.fit(X_train, y_train)
print("Logistic regression fitted.")
```

```python
#evaluate against baseline 
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)

models = {
    "Majority-class baseline": baseline,
    "Logistic regression": logit_model
}

results = []
for name, model in models.items():
    pred = model.predict(X_valid)
    if hasattr(model, "predict_proba"):
        prob = model.predict_proba(X_valid)[:, 1]
        auc = roc_auc_score(y_valid, prob)
    else:
        auc = np.nan
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_valid, pred),
        "Balanced accuracy": balanced_accuracy_score(y_valid, pred),
        "Precision": precision_score(y_valid, pred, zero_division=0),
        "Recall": recall_score(y_valid, pred, zero_division=0),
        "F1": f1_score(y_valid, pred, zero_division=0),
        "ROC-AUC": auc
    })

results_df = pd.DataFrame(results).set_index("Model")
results_df.round(3)
```

```python
#confusion matrix and classification report 
y_pred = logit_model.predict(X_valid)
y_prob = logit_model.predict_proba(X_valid)[:, 1]

print(classification_report(
    y_valid, y_pred,
    target_names=["Below 4.0 stars", "4.0+ stars"],
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_valid, y_pred,
    display_labels=["Below 4.0", "4.0 or higher"],
    values_format="d"
)
plt.title("Logistic Regression: Validation Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
#ROC curve 
RocCurveDisplay.from_predictions(y_valid, y_prob)
plt.plot([0, 1], [0, 1], linestyle="--", label="Chance")
plt.title("Logistic Regression: Validation ROC Curve")
plt.tight_layout()
plt.show()

print(f"Validation ROC-AUC: {roc_auc_score(y_valid, y_prob):.3f}")
```

```python
#model coefficients 
fitted_preprocessor = logit_model.named_steps["preprocessor"]
feature_names = fitted_preprocessor.get_feature_names_out()
coefficients = logit_model.named_steps["logistic_regression"].coef_[0]

coef_table = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coefficients,
    "odds_ratio": np.exp(coefficients)
})

print("Largest positive coefficients:")
display(coef_table.sort_values("coefficient", ascending=False).head(15).round(3))

print("Largest negative coefficients:")
display(coef_table.sort_values("coefficient").head(15).round(3))
```

```python
#checking predicted probabilities 
probability_summary = pd.DataFrame({
    "actual_high_rating": y_valid.to_numpy(),
    "predicted_probability_high_rating": y_prob
})

print(probability_summary["predicted_probability_high_rating"].describe().round(3))

plt.figure(figsize=(7, 4))
plt.hist(y_prob[y_valid.to_numpy() == 0], bins=25, alpha=0.6, label="Actual below 4.0")
plt.hist(y_prob[y_valid.to_numpy() == 1], bins=25, alpha=0.6, label="Actual 4.0+")
plt.xlabel("Predicted probability of 4.0+ stars")
plt.ylabel("Number of restaurants")
plt.title("Predicted probabilities by actual class")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score, log_loss,
    confusion_matrix, classification_report
)

# Predictions and probabilities
y_train_pred = logit_model.predict(X_train)
y_valid_pred = logit_model.predict(X_valid)

y_train_prob = logit_model.predict_proba(X_train)[:, 1]
y_valid_prob = logit_model.predict_proba(X_valid)[:, 1]

# McFadden's pseudo-R²
# Null model predicts the training-set majority-class probability
null_train_prob = np.full(len(y_train), y_train.mean())
null_valid_prob = np.full(len(y_valid), y_train.mean())

train_pseudo_r2 = 1 - (
    log_loss(y_train, y_train_prob, labels=[0, 1]) /
    log_loss(y_train, null_train_prob, labels=[0, 1])
)

valid_pseudo_r2 = 1 - (
    log_loss(y_valid, y_valid_prob, labels=[0, 1]) /
    log_loss(y_valid, null_valid_prob, labels=[0, 1])
)

# Model evaluation
evaluation = pd.DataFrame({
    "Training": [
        accuracy_score(y_train, y_train_pred),
        balanced_accuracy_score(y_train, y_train_pred),
        precision_score(y_train, y_train_pred, zero_division=0),
        recall_score(y_train, y_train_pred, zero_division=0),
        f1_score(y_train, y_train_pred, zero_division=0),
        roc_auc_score(y_train, y_train_prob),
        train_pseudo_r2
    ],
    "Validation": [
        accuracy_score(y_valid, y_valid_pred),
        balanced_accuracy_score(y_valid, y_valid_pred),
        precision_score(y_valid, y_valid_pred, zero_division=0),
        recall_score(y_valid, y_valid_pred, zero_division=0),
        f1_score(y_valid, y_valid_pred, zero_division=0),
        roc_auc_score(y_valid, y_valid_prob),
        valid_pseudo_r2
    ]
}, index=[
    "Accuracy",
    "Balanced Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC",
    "McFadden Pseudo-R²"
])

display(evaluation.round(4))

# Validation confusion matrix
print("Validation Confusion Matrix:")
print(confusion_matrix(y_valid, y_valid_pred))

print("\nValidation Classification Report:")
print(classification_report(
    y_valid,
    y_valid_pred,
    target_names=["Below 4.0 stars", "4.0+ stars"],
    zero_division=0
))
```

## KMEANS MODEL

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from scipy.sparse import issparse
```

```python
cluster_numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
cluster_categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=50
    ))
])

cluster_preprocessor = ColumnTransformer([
    ("numeric", cluster_numeric_pipeline, numeric_features),
    ("categorical", cluster_categorical_pipeline, categorical_features)
])

X_train_cluster = cluster_preprocessor.fit_transform(X_train)
X_valid_cluster = cluster_preprocessor.transform(X_valid)

print("Training clustering data:", X_train_cluster.shape)
print("Validation clustering data:", X_valid_cluster.shape)
```

```python
k_values = range(2, 9)

inertias = []
silhouette_scores = []

# Use a reproducible sample for silhouette calculations
rng = np.random.default_rng(6021)
sample_size = min(5000, X_train_cluster.shape[0])
sample_indices = rng.choice(
    X_train_cluster.shape[0],
    size=sample_size,
    replace=False
)

X_silhouette = X_train_cluster[sample_indices]

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=6021,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_train_cluster)

    inertias.append(kmeans.inertia_)

    sample_labels = cluster_labels[sample_indices]

    silhouette_scores.append(
        silhouette_score(
            X_silhouette,
            sample_labels
        )
    )

cluster_results = pd.DataFrame({
    "Number of clusters": list(k_values),
    "Inertia": inertias,
    "Silhouette Score": silhouette_scores
})

display(cluster_results.round(4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(
    list(k_values),
    inertias,
    marker="o"
)
axes[0].set_xlabel("Number of clusters (K)")
axes[0].set_ylabel("Inertia")
axes[0].set_title("K-means Elbow Method")
axes[0].grid(alpha=0.3)

axes[1].plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)
axes[1].set_xlabel("Number of clusters (K)")
axes[1].set_ylabel("Silhouette Score")
axes[1].set_title("K-means Silhouette Scores")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
best_k = list(k_values)[np.argmax(silhouette_scores)]

print("Selected number of clusters:", best_k)

kmeans_model = KMeans(
    n_clusters=best_k,
    random_state=6021,
    n_init=10
)

# Fit model on training data
train_clusters = kmeans_model.fit_predict(X_train_cluster)

# Assign validation restaurants to learned clusters
valid_clusters = kmeans_model.predict(X_valid_cluster)

print("K-means model fitted.")
print("Training silhouette score:",
      round(max(silhouette_scores), 4))
```

```python
train_cluster_counts = pd.Series(
    train_clusters
).value_counts().sort_index()

valid_cluster_counts = pd.Series(
    valid_clusters
).value_counts().sort_index()

cluster_sizes = pd.DataFrame({
    "Training": train_cluster_counts,
    "Validation": valid_cluster_counts
}).fillna(0).astype(int)

display(cluster_sizes)
```

```python
# Combine cluster assignments with actual outcomes

train_cluster_analysis = pd.DataFrame({
    "cluster": train_clusters,
    "high_rating": y_train.to_numpy()
})

valid_cluster_analysis = pd.DataFrame({
    "cluster": valid_clusters,
    "high_rating": y_valid.to_numpy()
})

# Training cluster rating distribution
train_rating_summary = (
    train_cluster_analysis
    .groupby("cluster")
    .agg(
        restaurant_count=("high_rating", "size"),
        high_rating_count=("high_rating", "sum"),
        high_rating_rate=("high_rating", "mean")
    )
)

# Validation cluster rating distribution
valid_rating_summary = (
    valid_cluster_analysis
    .groupby("cluster")
    .agg(
        restaurant_count=("high_rating", "size"),
        high_rating_count=("high_rating", "sum"),
        high_rating_rate=("high_rating", "mean")
    )
)

print("TRAINING CLUSTER RATING DISTRIBUTION")
display(train_rating_summary.round(3))

print("VALIDATION CLUSTER RATING DISTRIBUTION")
display(valid_rating_summary.round(3))
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    valid_rating_summary.index.astype(str),
    valid_rating_summary["high_rating_rate"]
)

plt.axhline(
    y=y_valid.mean(),
    linestyle="--",
    label="Overall validation high-rating rate"
)

plt.xlabel("Restaurant Cluster")
plt.ylabel("Proportion with 4.0+ Yelp Stars")
plt.title("High Yelp Rating Proportion by K-means Cluster")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Combine original training predictors and cluster assignments

cluster_profiles = X_train.copy()
cluster_profiles["cluster"] = train_clusters

# Numeric characteristics
numeric_profiles = (
    cluster_profiles
    .groupby("cluster")[numeric_features]
    .mean()
)

print("AVERAGE NUMERIC CHARACTERISTICS BY CLUSTER")
display(numeric_profiles.round(3))

# Categorical characteristics
for feature in categorical_features:

    print(f"\nMost common {feature} by cluster:")

    profile = (
        cluster_profiles
        .groupby("cluster")[feature]
        .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
    )

    display(profile)
```

## COMPARING LOGIT + KMEANS

```python
# Determine majority rating class within each training cluster

cluster_rating_mapping = (
    train_cluster_analysis
    .groupby("cluster")["high_rating"]
    .agg(lambda x: x.mode().iloc[0])
    .to_dict()
)

print("Cluster-to-rating mapping:")
print(cluster_rating_mapping)

# Assign predicted rating classes to validation clusters
kmeans_valid_predictions = pd.Series(valid_clusters).map(
    cluster_rating_mapping
).to_numpy()

print("\nK-means validation predictions generated.")
```

```python
kmeans_evaluation = pd.Series({
    "Accuracy": accuracy_score(y_valid, kmeans_valid_predictions),
    "Balanced Accuracy": balanced_accuracy_score(
        y_valid, kmeans_valid_predictions
    ),
    "Precision": precision_score(
        y_valid, kmeans_valid_predictions, zero_division=0
    ),
    "Recall": recall_score(
        y_valid, kmeans_valid_predictions, zero_division=0
    ),
    "F1 Score": f1_score(
        y_valid, kmeans_valid_predictions, zero_division=0
    )
})

display(kmeans_evaluation.round(4))

print("K-means validation classification report:")
print(classification_report(
    y_valid,
    kmeans_valid_predictions,
    target_names=["Below 4.0 stars", "4.0+ stars"],
    zero_division=0
))
```
